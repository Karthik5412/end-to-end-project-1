```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

### Reading **DF**

```python
df = spark.read.format('delta')\
    .load('abfss://bronze@etehub.dfs.core.windows.net/customers')

display(df)
```

### **Droping Unwanted**

```python
df = df.drop('_rescued_data')
```

### **Transformations**

```python
df = df.withColumn('domain',split(col("email"),'@')[1])
```

```python
df = df.withColumn('full_name', concat(col('first_name'), lit(' '), col('last_name')))
```

```python
df = df.drop('first_name','last_name')
```

```python
display(df)
```

### **Writing Df**

```python
df.write.format('delta').mode('append').save('abfss://silver@etehub.dfs.core.windows.net/customers')
```
