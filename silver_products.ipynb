```python
from pyspark.sql.functions import  *
from pyspark.sql.types import *
```

### **Reading table**

```python
df = spark.read.format('delta')\
    .load('abfss://bronze@etehub.dfs.core.windows.net/products')
```

```python
display(df)
```

### **Droping unwanted column**

```python
df = df.drop('_rescued_data')
```

### **Creating catalog functions**

```python
df.createOrReplaceTempView('temp_view')
```

**SQL**

```python
%sql
CREATE OR REPLACE FUNCTION `ete-catalog`.silver.discount(price INT)
RETURNS FLOAT
RETURN round(price * 0.9, 2);
```

```python
%sql
SELECT *, `ete-catalog`.silver.discount(price) AS final_price from temp_view
```

```python
df = df.withColumn('final_price', expr('`ete-catalog`.silver.discount(price)'))
```

```python
df.display()
```

**Python**

```python
%sql
CREATE OR REPLACE FUNCTION `ete-catalog`.silver.little(s STRING)
RETURNS STRING
LANGUAGE PYTHON
AS
$$
return s.lower() 
$$
```

```python
df_un = df.withColumn('little', expr("`ete-catalog`.silver.little(category)"))
```

```python
%sql
SELECT *,
`ete-catalog`.silver.little(category) AS LITTLE
FROM temp_view
```

```python
df_un.display()
```

### **Wrting df**

```python
df.write.format('delta')\
    .mode('append')\
    .save('abfss://silver@etehub.dfs.core.windows.net/products')
```
