```python
from pyspark.sql.functions import  *
from pyspark.sql.types import *
```

### **Reading table**

```python
df = spark.read.format('delta')\
    .load('abfss://bronze@etehub.dfs.core.windows.net/products')
```

```python
display(df)
```

### **Droping unwanted column**

```python
df = df.drop('_rescued_data')
```

### **Creating catalog functions**

```python
df.createOrReplaceTempView('temp_view')
```

**SQL**

```python
%sql
CREATE OR REPLACE FUNCTION `ete-catalog`.silver.discount(price INT)
RETURNS FLOAT
RETURN round(price * 0.9, 2);
```

```python
%sql
SELECT *, `ete-catalog`.silver.discount(price) AS final_price from temp_view
```
