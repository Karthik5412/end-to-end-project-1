### **Dlt tables**

```python
import dlt 
```

**stream table**

```python
@dlt.table(name = 'dim_products_stage')

def dim_products_stage() :
    df = spark.readStream.table('`ete-catalog`.silver.products')

    return df
```

**Streaming View**

```python
@dlt.view 

def product_view() :
    df = spark.readStream.table("Live.dim_products_stage")

    return df
```

**dim_products**

```python
dlt.create_streaming_table('dim_products')
```

```python
dlt.apply_changes(
    target= 'dim_products',
    source= 'Live.product_view',
    keys = ['product_id'],
    sequence_by = 'product_id',
    stored_as_scd_type = 2
)
```
